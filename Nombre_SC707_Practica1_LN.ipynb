## Creación de variables y vectorización

1. Resuelva la práctica de la presentación (Unidad 2.1) usando una lista con los valores en segundos, transfórmela en un array de numpy, escriba una función que reciba el array y retorne otro con los valores transformados en horas

```python
## Código de la solución
import numpy as np

segundos = [3600, 7200, 1800, 4500, 10000]

segundos_np = np.array(segundos)

def convertir_horas(arreglo):
    return arreglo/3600

horas = convertir_horas(segundos_np)
print(horas)
```

2. Resuelva el ejercicio de la presentación mediante un dataFrame. Cree un dataFrame con una fila y 5 columnas con los valores en segundos. Use apply() para generar una segunda fila con los valores transformados en horas.

```python
## Código de la solución
import pandas as pd

df = pd.DataFrame({
    "P1":[3600],
    "P2":[7200],
    "P3":[1800],
    "P4":[4500],
    "P5":[10000]
})

df.loc[1] = df.loc[0].apply(lambda x: x/3600)
print(df)
```

3. Complete el siguiente código para obtener la suma de los valores de cada columna y cada fila, guarde los valores en dos variables distintas

```python
## Código de la solución
import numpy as np

# Crear una matriz aleatoria 3x3
matrizAleatoria = np.random.randint(1, 10, (3, 3))

sumasColumnas = matrizAleatoria.sum(axis=0)
sumaFilas = matrizAleatoria.sum(axis=1)

print("Suma de columnas:", sumasColumnas)
print("Suma de filas:", sumaFilas)
```
